```python
import tensorflow as tf
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
```

```python
messages = [
    "Win a free prize now",
    "Claim your cash reward",
    "You won a lottery",
    "Free offer call now",
    "Hello how are you",
    "Meet me at college",
    "Please call me tomorrow",
    "Are you coming home",
    "Your prize is waiting",
    "Let us meet today"
]
```

```python
labels = np.array([1, 1, 1, 1, 0, 0, 0, 0, 1, 0])
```

```python
tokenizer = Tokenizer()
tokenizer.fit_on_texts(messages)
```

```python
sequences = tokenizer.texts_to_sequences(messages)
X = pad_sequences(sequences, maxlen=6)
```

```python
model = Sequential([
    Embedding(100, 16, mask_zero=True),
    LSTM(16),
    Dense(1, activation="sigmoid")
])
```

```python
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

```python
model.fit(X, labels, epochs=30, verbose=0)
```

```python
test_messages = [
    "Win a free prize",
    "Hello meet me today"
]
```

```python
test_sequences = tokenizer.texts_to_sequences(test_messages)
test_X = pad_sequences(test_sequences, maxlen=6)
```

```python
predictions = model.predict(test_X, verbose=0)
```

```python
for message, prediction in zip(test_messages, predictions):
    result = "Spam" if prediction[0] >= 0.5 else "Not Spam"
    print(message, ":", result)
```
